```python
orders = pd.read_excel("BD.xlsx", sheet_name="orders")
order_statuses = pd.read_excel("BD.xlsx", sheet_name="order_statuses")
menu = pd.read_excel("BD.xlsx", sheet_name="menu")
order_items = pd.read_excel("BD.xlsx", sheet_name="order_items")
client_types = pd.read_excel("BD.xlsx", sheet_name="client_types")
cashiers = pd.read_excel("BD.xlsx", sheet_name="cashiers")
```

```python
# Загрузить данные из csv в бд
from sqlalchemy import create_engine, inspect
import pandas as pd

engine = create_engine('postgresql://user:password@localhost:port/database')

for i in ["client_types", "order_statuses", "cashiers", "menu", "orders", "order_items"]:
    inspector = inspect(engine)
    columns = [col["name"] for col in inspector.get_columns(i)]

    df = pd.read_csv(f"{i}.csv")
    if len(df.columns) == len(columns):
        df.columns = columns
        df.to_sql(f'{i}', engine, if_exists='append', index=False)
    else:
        print("Ошибка")
        break

```

```python
# Сохранить sql таблицы в csv (бэкап)
for i in ["client_types", "order_statuses", "cashiers", "menu", "orders", "order_items"]:
    df = pd.read_sql_table(i, engine)
    df.to_csv(f"{i}.csv")
```
